```python
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))
operation = str(input("What operation? A = addition, B = subtraction, C = Multiplication, D = division: "))

def add(num1, num2):
  sum = num1 + num2
  return sum

def subtract(num1, num2):
  difference = num1 - num2
  return difference

def multiply(num1, num2):
  product = num1 * num2
  return product

def divide(num1, num2):
  if num1 <= 0:
    num1 = int(input("First number must not be zero: "))
  if num2 <= 0:
    num2 = int(input("Second number must not be zero: "))
  quotient = num1 / num2
  return quotient

if operation == "A":
  print(add(num1, num2))
elif operation == "B":
  print(subtract(num1, num2))
elif operation == "C":
  print(multiply(num1, num2))
elif operation == "D":
  print(divide(num1, num2))

```
